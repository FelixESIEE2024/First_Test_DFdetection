```python
import os
from pathlib import Path
import shutil
import subprocess

repo_dir = Path(r"C:\Users\conqu\Videos\Detection_Deep_Fake\Geco\GeoCalib\GeoCalib")
source_calib = repo_dir / "calib"

object_centric_dir = Path(
    #r"C:\Users\conqu\Videos\Detection_Deep_Fake\Repo_1st_test\Implementaiton_1\dataset\ai\Geco\Gen_CogVideoX_2b\Gen_CogVideoX_2b\object_centric_prompts\oue"
    r"C:\Users\conqu\Videos\Detection_Deep_Fake\Répo_1st_test\Implementaiton_1\dataset\ai\Geco\deform"
)

os.chdir(repo_dir)

for folder in object_centric_dir.iterdir():
    if not folder.is_dir():
        continue

    frame = folder
    destination = folder / "calib"

    print(f"Traitement du dossier : {frame}")

    # Supprime l'ancien calib généré par intra.py
    if source_calib.exists():
        shutil.rmtree(source_calib)

    # Exécute intra.py sur les frames du dossier courant
    subprocess.run(
        ["python", "intra.py", "--path_frame", str(frame)],
        check=True
    )

    # Vérifie que intra.py a bien généré calib
    if not source_calib.exists():
        raise FileNotFoundError(f"Le dossier calib n'a pas été généré pour {frame}")

    # Supprime calib dans le dossier cible s'il existe déjà
    if destination.exists():
        shutil.rmtree(destination)

    # Copie le résultat dans le dossier courant
    shutil.copytree(source_calib, destination)

    print(f"Calib copié vers : {destination}")
```
